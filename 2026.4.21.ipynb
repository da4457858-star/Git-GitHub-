```python
n = int(input("number n : "))
sum = 1
for num in range(1,n+1):
    sum *= num
print("result=",sum)
```

```python
for i in range(1,10):
    for j in range(1,10):
        result = i * j
        print("%d*%d=%-3d" % (i,j,result), end=" ")
    print() 
```

```python
for i in range(1,10):
    for j in range(1,10):
        if j <= i :
            print("*",end="")
    print()
```

```python
for i in range(10,1,-1):
    for j in range(1,i):
        if j <= i :
            print("*",end="")
    print()
```

```python
for i in range(1,10):
    for j in range(1,9-i+1):
            print(" ",end="")
    for j in range(1,i+1):
            print("*",end="")
    print()
```

```python
answer = 30 
guess = 0
while guess != answer:
    guess = int(input("guess 1-100 = "))
    if guess > answer :
        print("guess small")
    elif guess < answer :
        print("guess big")
    else:
        print("right")


```
